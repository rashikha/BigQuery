````python
import sys
import requests


def download_file_from_google_drive(id, destination):
    URL = "https://docs.google.com/uc?export=download&confirm=1"

    session = requests.Session()

    response = session.get(URL, params={"id": id}, stream=True)
    token = get_confirm_token(response)

    if token:
        params = {"id": id, "confirm": token}
        response = session.get(URL, params=params, stream=True)

    save_response_content(response, destination)


def get_confirm_token(response):
    for key, value in response.cookies.items():
        if key.startswith("download_warning"):
            return value

    return None


def save_response_content(response, destination):
    CHUNK_SIZE = 32768

    with open(destination, "wb") as f:
        for chunk in response.iter_content(CHUNK_SIZE):
            if chunk:  # filter out keep-alive new chunks
                f.write(chunk)
````

````python
def add_driver_options(options):
    """
    Add configurable options
    """
    chrome_options = Options()
    for opt in options:
        chrome_options.add_argument(opt)
    return chrome_options

def initialize_driver():
    """
    Initialize the web driver
    """
    driver_config = {
        "executable_path": binary_path,
        "options": [
            "--headless",
            "--no-sandbox",
            "--start-fullscreen",
            "--allow-insecure-localhost",
            "--disable-dev-shm-usage",
        ],
    }
    options = add_driver_options(driver_config["options"])
    driver = webdriver.Chrome(
        executable_path=driver_config["executable_path"], options=options
    )
    return driver
````

````python
def timeStampToSeconds(x):
    hour = datetime.datetime.strptime(x, '%Y-%m-%d %H:%M:%S').strftime("%-H")
    minute = datetime.datetime.strptime(x, '%Y-%m-%d %H:%M:%S').strftime("%-M")
    second = datetime.datetime.strptime(x, '%Y-%m-%d %H:%M:%S').strftime("%-S")
    calcSeconds = (float(hour)-9)*60*60 + float(minute)*60 + float(second)

    return (23400 - calcSeconds)
````

````python
# Auto-scaling TPU matmul pinger (use measured baseline to pick repeats)
import threading, time, sys, os, traceback
import numpy as np
import jax, jax.numpy as jnp

HEARTBEAT_FILE = "/tmp/tpu_pinger_hb_autoscale.txt"
_tpu_stop_event = threading.Event()
_tpu_thread_ref = None

def _write_hb(ts=None):
    try:
        with open(HEARTBEAT_FILE, "w") as f:
            f.write(str(ts or time.time()))
    except Exception as e:
        print("[TPU pinger] HB write failed:", e, flush=True)

def make_matmul_pmap_fn(M, local_batch, repeats):
    def body_fn(x):
        def body(i, val):
            return jnp.matmul(val, val)
        return jax.lax.fori_loop(0, repeats, body, x)
    return jax.pmap(jax.jit(body_fn), axis_name='i')

def autoscale_matmul_pinger(
        interval=16,                 # seconds between ping starts (measured from start)
        target_active_seconds=13,  # desired active seconds per ping
        # baseline measured input (fill with your measured baseline)
        baseline_M=512,
        baseline_local_batch=32,
        baseline_repeats=50000,
        baseline_measured_seconds=6.0,
        # safety caps
        max_repeats=120000,
        safety_local_batch_cap=128
    ):
    """
    Auto-scale repeats_per_ping from your measured baseline. Defaults tuned to your measurement.
    """
    print("[TPU pinger-auto] starting; interval", interval, "target_active_seconds", target_active_seconds, flush=True)

    # wait for devices
    devices = None
    for _ in range(60):
        try:
            devices = jax.devices("tpu")
            if devices:
                break
        except Exception:
            pass
        time.sleep(1)
    if not devices:
        print("[TPU pinger-auto] No TPU devices found; exiting.", flush=True)
        return
    num_devices = len(devices)
    print("[TPU pinger-auto] devices:", devices, flush=True)

    # Choose working params. Start with baseline M/local_batch (your measurements).
    M = baseline_M
    local_batch = min(baseline_local_batch, safety_local_batch_cap)

    # Compute repeats needed to hit target (linear scaling)
    # repeats_target = baseline_repeats * (target / baseline_measured_seconds)
    repeats_target = int(baseline_repeats * (float(target_active_seconds) / float(baseline_measured_seconds)))
    repeats_target = max(1, repeats_target)
    repeats_target = min(repeats_target, max_repeats)  # clamp

    print(f"[TPU pinger-auto] baseline: M={baseline_M}, local_batch={baseline_local_batch}, "
          f"repeats={baseline_repeats}, measured_s={baseline_measured_seconds}", flush=True)
    print(f"[TPU pinger-auto] chosen: M={M}, local_batch={local_batch}, repeats_target={repeats_target}", flush=True)

    # Prepare input array: (num_devices, local_batch, M, M)
    try:
        arr = np.random.randn(num_devices, local_batch, M, M).astype(np.float32)
    except Exception as e:
        print("[TPU pinger-auto] Failed to allocate array with these params:", e, flush=True)
        print("[TPU pinger-auto] Try reducing local_batch or M", flush=True)
        return

    # Build and run one calibrating run (may compile)
    try:
        print("[TPU pinger-auto] performing calibration warm-run (may take time)...", flush=True)
        pmap_fn = make_matmul_pmap_fn(M, local_batch, repeats_target)
        t0 = time.time()
        out = pmap_fn(jnp.array(arr))
        _ = jax.device_get(out)
        dur = time.time() - t0
        print(f"[TPU pinger-auto] calibration run done, dur={dur:.3f}s", flush=True)
    except Exception as e:
        print("[TPU pinger-auto] calibration failed:", e, flush=True)
        traceback.print_exc()
        # fallback: try with smaller repeats
        try:
            fallback_repeats = max(1, repeats_target // 4)
            print("[TPU pinger-auto] trying fallback smaller repeats:", fallback_repeats, flush=True)
            pmap_fn = make_matmul_pmap_fn(M, local_batch, fallback_repeats)
            t0 = time.time()
            out = pmap_fn(jnp.array(arr))
            _ = jax.device_get(out)
            dur = time.time() - t0
            print(f"[TPU pinger-auto] fallback run dur={dur:.3f}s", flush=True)
            repeats_target = fallback_repeats
        except Exception as e2:
            print("[TPU pinger-auto] fallback also failed:", e2, flush=True)
            return

    # If resulting dur is much shorter than target, scale repeats proportionally
    if dur > 0:
        scaled_repeats = int(repeats_target * (target_active_seconds / dur))
        scaled_repeats = min(scaled_repeats, max_repeats)
        if scaled_repeats != repeats_target:
            print(f"[TPU pinger-auto] scaling repeats from {repeats_target} -> {scaled_repeats} (target {target_active_seconds}s)", flush=True)
            repeats_target = scaled_repeats
            try:
                pmap_fn = make_matmul_pmap_fn(M, local_batch, repeats_target)
            except Exception as e:
                print("[TPU pinger-auto] rebuild pmap failed for scaled repeats:", e, flush=True)
                return

    print(f"[TPU pinger-auto] final repeats_per_ping={repeats_target}. Starting loop (interval={interval}s).", flush=True)

    # Main loop
    iter_no = 0
    while not _tpu_stop_event.is_set():
        iter_no += 1
        start_t = time.time()
        try:
            out = pmap_fn(jnp.array(arr))
            _ = jax.device_get(out)
            elapsed = time.time() - start_t
            print(f"[TPU pinger-auto] iter={iter_no} elapsed={elapsed:.2f}s repeats={repeats_target} local_batch={local_batch} M={M}", flush=True)
        except Exception as e:
            print("[TPU pinger-auto] ERROR during pmap:", e, flush=True)
            traceback.print_exc()
        _write_hb()
        # sleep until next cycle measured from start
        to_sleep = max(0.0, interval - (time.time() - start_t))
        slept = 0.0
        while slept < to_sleep and not _tpu_stop_event.is_set():
            time.sleep(1.0)
            slept += 1.0

    print("[TPU pinger-auto] stopping.", flush=True)

# Start the pinger thread using your measured baseline:
_tpu_thread_ref = threading.Thread(
    target=autoscale_matmul_pinger,
    kwargs=dict(
        interval=16,
        target_active_seconds=13,
        baseline_M=512,
        baseline_local_batch=32,
        baseline_repeats=50000,
        baseline_measured_seconds=6.0,
        max_repeats=120000
    ),
    daemon=True
)
_tpu_thread_ref.start()
print("Started autoscale TPU pinger; heartbeat:", HEARTBEAT_FILE, flush=True)

# stop helper
def stop_tpu_pinger_autoscale(timeout=10):
    _tpu_stop_event.set()
    if _tpu_thread_ref is not None:
        _tpu_thread_ref.join(timeout=timeout)
    print("TPU pinger-autoscale stopped.", flush=True)
````

````python
import base64
import pandas as pd
from IPython.display import HTML

def create_download_link( df, title = "Download CSV file", filename = "data.csv"):
    csv = df.to_csv()
    b64 = base64.b64encode(csv.encode())
    payload = b64.decode()
    html = '<a download="{filename}" href="data:text/csv;base64,{payload}" target="_blank">{title}</a>'
    html = html.format(payload=payload,title=title,filename=filename)
    return HTML(html)
````
